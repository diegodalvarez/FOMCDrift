# Examining Raw and Volatility Hedged Returns

```python
df_ordered = (df_open
    .merge(right = df_meeting_id, how = "inner", on = ["fomc_date"])
    .assign(
        time_to_meet = lambda x: x.date - x.fomc_date,
        time         = lambda x: x.date.dt.time))
```

```python
def _get_rtn(df: pd.DataFrame, vol_days: int = 10, threshold: int = 50) -> pd.DataFrame: 

    df_wider = (df
        .pivot(index = "time_to_meet", columns = "meeting_id", values = "value")
        .ffill()
        .diff()
        .fillna(0))

    df_vol_diff = (df_wider
        .apply(lambda x: x / x.ewm(span = vol_days, adjust = False).std())
        .apply(lambda x: np.where(np.abs(x) > threshold, np.nan, x))
        .reset_index()
        .melt(id_vars = "time_to_meet", value_name = "vol_diff")
        .dropna())

    df_px_diff = (df_wider
        .reset_index()
        .melt(id_vars = "time_to_meet", value_name = "px_diff"))

    df_out = (df_vol_diff
        .merge(right = df_px_diff, how = "inner", on = ["time_to_meet", "meeting_id"]))

    return df_out

df_rtn = (df_ordered
    .groupby("ticker")
    .apply(_get_rtn)
    .reset_index()
    .drop(columns = ["level_1"]))
```

```python
df_hedged = (df_ordered
    .merge(right = df_rtn, how = "inner", on = ["ticker", "time_to_meet", "meeting_id"]))
```

df_px_diff = (df_ordered
    [["value", "new_file", "date"]]

```python
tickers   = df_hedged.ticker.drop_duplicates().sort_values().to_list()
fig, axes = plt.subplots(ncols = len(tickers) // 2, nrows = 2, figsize = (20,12))

for ticker, ax in zip(tickers, axes.flatten()): 

    (df_hedged
        .loc[lambda x: x.ticker == ticker]
        .pivot(index = "time_to_meet", columns = "meeting_id", values = "px_diff")
        .fillna(0)
        .cumsum()
        .plot(
            legend = False, 
            alpha  = 0.7,
            title  = ticker,
            ylabel = "Cuml. PnL",
            ax     = ax,
            xlabel = "Time to Meeting"))

fig.suptitle("Cumulative PX PnL of Treasury Futures Around FOMC Dates")
plt.tight_layout()
```

# Defining Trade Windows

```python
df_hedged_times = (df_hedged
    [["date", "time_to_meet", "time", "fomc_date"]]
    .dropna()
    .assign(
        day_time_to_meet = lambda x: pd.to_datetime(x.fomc_date.dt.date) - pd.to_datetime(x.date.dt.date),
        days_to_meeting  = lambda x: x.day_time_to_meet.dt.days))
```

```python
start_time = dt.time(hour = 9, minute = 30)
end_time   = dt.time(hour = 12 + 3, minute = 30)

df_day_before = (df_hedged_times
    .loc[lambda x: x.days_to_meeting == 1]
    .loc[lambda x: (start_time <= x.time) & (x.time <= end_time)]
    .assign(
        zone = "zone1",
        name = "day_before"))
```

```python
start_time = dt.time(hour = 9, minute = 30)
end_time   = dt.time(hour = 12 + 1, minute = 0)

df_premeeting = (df_hedged_times
    .loc[lambda x: x.days_to_meeting == 0]
    .loc[lambda x: (start_time <= x.time) & (x.time <= end_time)]
    .assign(
        zone = "zone2",
        name = "pre_meeting"))
```

```python
start_time = dt.time(hour = 12 + 2, minute = 30)
end_time   = dt.time(hour = 12 + 3, minute = 30)

df_postmeeting = (df_hedged_times
    .loc[lambda x: x.days_to_meeting == 0]
    .loc[lambda x: (start_time <= x.time) & (x.time <= end_time)]
    .assign(
        zone = "zone3",
        name = "post_meeting"))
```

```python
start_time = dt.time(hour = 9, minute = 30)
end_time   = dt.time(hour = 12 + 3, minute = 30)

df_after = (df_hedged_times
    .loc[lambda x: x.days_to_meeting == -1]
    .loc[lambda x: (start_time <= x.time) & (x.time <= end_time)]
    .assign(
        zone = "zone4",
        name = "day_after"))
```

```python
np.random.seed(42)
nums = np.random.choice(np.arange(69, 148), size=20, replace=False)

df_wider = (
    df_hedged
    .loc[lambda x: x.ticker == x.ticker.min()]
    .loc[lambda x: x.meeting_id.isin(nums)]
    .pivot(
        index   = "time_to_meet",
        columns = "meeting_id",
        values  = "vol_diff")
    .fillna(0)
    .apply(lambda x: np.cumsum(x)))
```

```python
min_max_dict = (df_wider
    .reset_index()
    .melt(id_vars = "time_to_meet")
    [["value"]]
    .agg(["min", "max"])
    .value
    .to_dict())

min_val, max_val = (1.1 * min_max_dict["min"]), (1.1 * min_max_dict["max"])
```

```python
df_plot_wider = df_wider.copy()

x_labels = df_plot_wider.index
x = df_plot_wider.index.total_seconds()

windows = [
    (df_day_before, "red", "Day before"),
    (df_premeeting, "green", "Pre-meeting"),
    (df_postmeeting, "orange", "Post-meeting"),
    (df_after, "purple", "Next day"),
]

fig, axes = plt.subplots(figsize=(10, 6))

legend_handles = []

for df, color, label in windows:

    window_x = (
        df.time_to_meet
        .drop_duplicates()
        .sort_values()
        .dt.total_seconds()
        .to_numpy()
    )

    axes.fill_between(
        x=window_x,
        y1=min_val,
        y2=max_val,
        color=color,
        alpha=0.2
    )

    legend_handles.append(
        Patch(
            facecolor=color,
            alpha=0.2,
            label=label
        )
    )

df_plot_wider.index = x

df_plot_wider.plot(
    ax=axes,
    legend=False,
    colormap="Blues"
)

idx = np.linspace(0, len(x) - 1, 8).astype(int)

axes.set_xticks(x[idx])
axes.set_xticklabels(
    x_labels[idx],
    rotation=45,
    ha="right"
)

axes.axvline(
    x=0,
    color="black",
    linestyle="--",
    linewidth=1.5
)

legend_handles.append(
    Patch(
        facecolor="black",
        alpha=1,
        label="FOMC Meeting"
    )
)

axes.legend(handles=legend_handles)

axes.set_ylabel("Cumulative PnL")
axes.set_xlabel("Time to FOMC Meeting (Pythonic)")
fig.suptitle(
    "Cumulative Returns of Some Sample Meetings with Trade Windows\n"
    "Volatility Hedged TN"
)

plt.show()
```

# Naive Momentum Backtest

```python
df_zones = (pd.concat([
    df_day_before,
    df_premeeting,
    df_postmeeting,
    df_after]))
```

```python
df_zoned_rtn = (df_hedged
    .merge(right = df_zones, how = "inner", on = ["date", "time_to_meet", "fomc_date", "time"]))
```

```python
df_avg_per_zone = (df_zoned_rtn
    [["ticker", "time", "zone", "name", "vol_diff"]]
    .groupby(["ticker", "zone", "name", "time"])
    .agg("mean")
    .reset_index())
```

```python
zone_renamer = (df_avg_per_zone
    [["zone", "name"]]
    .drop_duplicates()
    .assign(name = lambda x: x.name.str.replace("_", " ").str.capitalize())
    .set_index("zone")
    .name
    .to_dict())
```

```python
zones     = df_avg_per_zone.zone.drop_duplicates().sort_values().to_list()
fig, axes = plt.subplots(ncols = len(zones) // 2, nrows = len(zones) // 2, figsize = (20,10)) 

for zone, ax in zip(zones, axes.flatten()): 

    (df_avg_per_zone
        .loc[lambda x: x.zone == zone]
        .rename(columns = {"ticker": ""})
        .pivot(index = "time", columns = "", values = "vol_diff")
        .cumsum()
        .plot(
            ax    = ax,
            title = zone_renamer[zone],
            ylabel = "Cuml. Ret."))

fig.suptitle("Cumulative PnL of Volatility Targeted Returns During Each Defined Trading Zone")
plt.show()
```

Start by making a very naive approach of taking a position corresponding ending value of each zone and apply it. 

```python
def _get_max(df: pd.DataFrame) -> pd.DataFrame: return df.loc[lambda x: x.time == x.time.max()]

df_signal = (df_avg_per_zone
    .set_index("time")
    .groupby(["ticker", "name"])
    .apply(lambda x: x.sort_index().vol_diff.cumsum())
    .reset_index()
    .groupby(["ticker", "name"])
    .apply(_get_max)
    .reset_index()
    .drop(columns = ["level_2", "time"])
    .rename(columns = {"vol_diff": "signal"}))
```

```python
df_signal_rtn = (df_zoned_rtn
    [["fomc_date", "time_to_meet", "ticker", "name", "meeting_id", "px_diff", "vol_diff"]]
    .melt(id_vars = ["fomc_date", "time_to_meet", "ticker", "name", "meeting_id"])
    .merge(right = df_signal, how = "inner", on = ["ticker", "name"])
    .assign(signal_rtn = lambda x: np.sign(x.signal) * x.value))
```

```python
df_sharpe = (df_signal_rtn
    [["ticker", "name", "variable", "signal_rtn"]]
    .groupby(["ticker", "name", "variable"])
    .agg(lambda x: x.mean() / x.std() * np.sqrt(1440 * 252))
    .reset_index()
    .pivot(index = ["ticker", "name"], columns = "variable", values = "signal_rtn"))
```

```python
df_tmp = (df_signal_rtn
    .assign(group_var = lambda x: x.ticker + "-" + x.name + "-" + x.meeting_id.astype(str) + "-" + x.variable)
    .loc[lambda x: x.group_var == x.group_var.min()])
```

```python
def _get_cuml_pnl(df: pd.DataFrame) -> pd.DataFrame: 
    
    df_cuml = (df
        .set_index("time_to_meet")
        .sort_index()
        [["signal_rtn"]]
        .cumsum())

(df_tmp
    .groupby("group_var")
    .apply(_get_cuml_pnl))
```
