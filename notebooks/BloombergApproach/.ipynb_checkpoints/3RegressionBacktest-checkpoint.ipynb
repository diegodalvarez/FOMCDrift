# Regression Backtest

```python
import os
import numpy as np
import pandas as pd
import statsmodels.api as sm
import matplotlib.pyplot as plt
from   matplotlib.ticker import FuncFormatter
```

```python
from tqdm import tqdm
tqdm.pandas()
```

# Get Data

```python
root_path = os.path.abspath(os.path.join(os.getcwd(), ".."))
data_path = os.path.join(root_path, "data")
raw_path  = os.path.join(data_path, "RawData")
tsy_path  = os.path.join(raw_path, "TreasuryFutures.parquet")

process_path = os.path.join(data_path, "ProcessedData")
date_path    = os.path.join(process_path, "OrderedDates.parquet")
```

```python
df_tsy   = pd.read_parquet(path = tsy_path, engine = "pyarrow")
df_dates = pd.read_parquet(path = date_path, engine = "pyarrow")
```

```python
processed_path = os.path.join(process_path, "ProcessedSentimentData.parquet")
df_zscore      = (pd.read_parquet(
    path = processed_path, engine = "pyarrow")
    [["date", "security", "lag_zscore"]])
```

```python
df_zscore_tmp = (df_dates.query(
    "day == day.min()")
    [["date", "event"]].
    drop_duplicates().
    merge(right = df_zscore, how = "inner", on = ["date"]))
```

```python
df_fomc_rtn = (df_tsy.assign(
    date = lambda x: pd.to_datetime(x.date).dt.date).
    merge(right = df_dates, how = "inner", on = ["date"]).
    assign(security = lambda x: x.security.str.split("1").str[0].str.strip()).
    pivot(index = "day", columns = ["event", "security"], values = "PX_bps").
    cumsum().
    ffill().
    reset_index().
    melt(id_vars = [("day", "")], value_name = "rtn").
    rename(columns = {("day", ""): "day"}).
    query("day == day.max()").
    drop(columns = ["day"]))
```

# Single Period Regression

```python
df_panel = (df_zscore_tmp.pivot(
    index = ["date", "event"], columns = "security", values = "lag_zscore").
    reset_index().
    merge(right = df_fomc_rtn, how = "inner", on = ["event"]))
```

```python
def _get_ols_pred(df: pd.DataFrame) -> pd.DataFrame: 

    fitted_values = (sm.OLS(
        endog = df.rtn,
        exog  = sm.add_constant(df[["APUSISGF", "APUSSPGF", "APUSTYGF", "APUSXRGF", "BENLPFED"]])).
        fit().
        fittedvalues)

    df_out = (df.assign(
        y_pred     = fitted_values,
        signal_rtn = lambda x: np.sign(x.y_pred) * x.rtn))

    return df_out

df_ols_rtn = (df_panel.fillna(
    0).
    groupby("security").
    apply(_get_ols_pred, include_groups = False).
    reset_index().
    drop(columns = ["level_1"]))
```

```python
def _get_logit_pred(df: pd.DataFrame) -> pd.DataFrame: 

    df_tmp = (df.assign(
        rtn = lambda x: np.where(x.rtn > 0, 1, 0)))
    
    fitted_values = (sm.Logit(
        endog = df_tmp.rtn,
        exog  = sm.add_constant(df_tmp[["APUSISGF", "APUSSPGF", "APUSTYGF", "APUSXRGF", "BENLPFED"]])).
        fit().
        fittedvalues)

    df_out = (df.assign(
        y_pred     = fitted_values,
        signal_rtn = lambda x: np.sign(x.y_pred) * x.rtn))

    return df_out

df_logit_rtn = (df_panel.fillna(0).groupby(
    "security").
    apply(_get_logit_pred, include_groups = False).
    reset_index())
```

```python
df_signal = (pd.concat([
    df_logit_rtn[["security", "event", "y_pred"]].assign(group = "logit"),
    df_ols_rtn[["security", "event", "y_pred"]].assign(group = "OLS")]))
```

```python
df_signal_rtn = (df_tsy.assign(
    date     = lambda x: pd.to_datetime(x.date).dt.date,
    security = lambda x: x.security.str.split("1").str[0].str.strip()).
    merge(right = df_dates, how = "inner", on = ["date"])
    [["date", "security", "event", "PX_bps"]].
    merge(right = df_signal, how = "inner", on = ["event", "security"]).
    assign(signal_rtn = lambda x: np.sign(x.y_pred) * x.PX_bps))
```

```python
groups    = df_signal_rtn.group.drop_duplicates().sort_values().to_list()
fig, axes = plt.subplots(ncols = 2, figsize = (20,6))

renamer = {
    "OLS"  : "OLS Regression",
    "logit": "Logistic Regression"}

for group, ax in zip(groups, axes.flatten()): 

    (df_signal_rtn.query(
        "group == @group")
        [["date", "security", "signal_rtn"]].
        drop_duplicates().
        rename(columns = {"security": ""}).
        pivot(index = "date", columns = "", values = "signal_rtn").
        cumsum().
        plot(
            ax     = ax,
            ylabel = "Cuml. Basis Points",
            title  = renamer[group]))

fig.suptitle("Cumulative Basis Points of Trading Single Period Regression")
plt.show()
```

```python
(df_signal_rtn[
    ["security", "group", "signal_rtn"]].
    groupby(["security", "group"]).
    agg(lambda x: x.mean() / x.std() * np.sqrt(252)).
    reset_index().
    pivot(index = "group", columns = "security", values = "signal_rtn"))
```

```python
(df_signal_rtn[
    ["date", "security", "group", "signal_rtn"]].
    drop_duplicates().
    pivot(index = ["date", "security"], columns = "group", values = "signal_rtn").
    reset_index().
    drop(columns = ["date"]).
    groupby("security").
    agg("corr").
    query("logit != 1").
    reset_index().
    set_index("security")
    [["logit"]].
    T)
```

```python
df_mean_port = (df_signal_rtn[
    ["date", "security", "signal_rtn"]].
    drop_duplicates().
    groupby(["date", "security"]).
    agg("mean").
    reset_index())
```

```python
(df_mean_port.pivot(
    index = "date", columns = "security", values = "signal_rtn").
    cumsum().
    plot())
```

```python
(df_mean_port.drop(
    columns = ["date"]).
    groupby(["security"]).
    agg(lambda x: x.mean() / x.std() * np.sqrt(252)).
    T)
```

# Multi-Period Regression

```python
def _cumulate_rtn(df: pd.DataFrame) -> pd.DataFrame: 

    df_out = (df.sort_values(
        "date")
        [["date", "security", "PX_bps", "event", "group"]].
        set_index(["date", "security", "event", "group"]).
        cumsum().
        reset_index().
        query("date == date.max()"))

    return df_out

df_tsy_prep = (df_tsy.assign(
    date = lambda x: pd.to_datetime(x.date).dt.date).
    merge(right = df_dates, how = "inner", on = ["date"]).
    assign(
        group    = lambda x: np.where(x.day <= 0, "before", "after"),
        security = lambda x: x.security.str.split("1").str[0].str.strip(),
        name     = lambda x: x.security + " " + x.event.astype(str) + " " + x.group).
    groupby("name").
    apply(_cumulate_rtn).
    reset_index().
    reset_index(drop = True))
```

```python
df_combined = (df_zscore_tmp.pivot(
    index = ["event"], columns = "security", values = "lag_zscore").
    reset_index().
    merge(right = df_tsy_prep, how = "inner", on = ["event"]).
    drop(columns = ["level_1"]).
    assign(name = lambda x: x.security + " " + x.group))
```

```python
def _run_regression(df: pd.DataFrame) -> pd.DataFrame: 

    df_dates = df[["date", "event"]].drop_duplicates()

    df_tmp = (df.set_index(
        "date")
        [["APUSISGF", "APUSSPGF", "APUSTYGF", "APUSXRGF", "BENLPFED", "PX_bps"]].
        fillna(0))

    exog_cols = [col for col in df_tmp.columns.to_list() if col != "PX_bps"]
    ols_pred  = (sm.OLS(
        endog = df_tmp.PX_bps,
        exog  = sm.add_constant(df_tmp[exog_cols])).
        fit().
        fittedvalues)

    df_tmp = df_tmp.assign(ols_PX_bps = lambda x: np.where(x.PX_bps > 0, 1, 0))
    
    logit_pred = (sm.Logit(
        endog = df_tmp.ols_PX_bps,
        exog  = sm.add_constant(df_tmp[exog_cols])).
        fit().
        fittedvalues)

    security, group = df.name.split(" ")
    
    df_out = (df_tmp.assign(
        ols_pred   = ols_pred,
        logit_pred = logit_pred,
        security   = security,
        group      = group).
        merge(right = df_dates, how = "inner", on = ["date"]))

    return df_out

df_signal = (df_combined.groupby(
    "name").
    apply(_run_regression, include_groups = False).
    reset_index(drop = True))
```

```python
df_signal = (df_signal[
    ["ols_pred", "security", "group", "event", "logit_pred"]].
    melt(id_vars = ["security", "group", "event"], var_name = "model", value_name = "signal"))
```

```python
df_signal_rtn = (df_tsy.assign(
    date = lambda x: pd.to_datetime(x.date).dt.date).
    merge(right = df_dates, how = "inner", on = ["date"]).
    assign(
        security = lambda x: x.security.str.split("1").str[0].str.strip(),
        group    = lambda x: np.where(x.day <= 0, "before", "after"))
    [["date", "group", "security", "PX_bps", "event"]].
    merge(right = df_signal, how = "inner", on = ["event", "security", "group"]).
    drop(columns = ["group"]).
    drop_duplicates().
    assign(signal_rtn = lambda x: np.sign(x.signal) * x.PX_bps))
```

```python
models    = df_signal_rtn.model.drop_duplicates().sort_values().to_list()
fig, axes = plt.subplots(ncols = len(models), figsize = (20,6))

renamer = {
    "ols_pred"  : "OLS",
    "logit_pred": "Logistic"}

for model, ax in zip(models, axes.flatten()):

    (df_signal_rtn.query(
        "model == @model")
        [["date", "security", "signal_rtn"]].
        drop_duplicates().
        rename(columns = {"security": ""}).
        pivot(index = "date", columns = "", values = "signal_rtn").
        cumsum().
        plot(
            ax      = ax,
            ylabel  = "Cuml. Basis Points",
            title   = renamer[model]))

fig.suptitle("Multi-Preiod Regression")
plt.tight_layout()
```

```python
(df_signal_rtn[
    ["security", "model", "signal_rtn"]].
    groupby(["security", "model"]).
    agg(lambda x: x.mean() / x.std() * np.sqrt(252)).
    reset_index().
    pivot(index = "model", columns = "security", values = "signal_rtn"))
```

```python
df_model_wider = (df_signal_rtn[
    ["date", "security", "signal_rtn", "model"]].
    drop_duplicates().
    pivot(index = ["date", "security"], columns = "model", values = "signal_rtn").
    reset_index())
```

```python
(df_model_wider.drop(
    columns = ["date"]).
    groupby("security").
    agg("corr").
    query("ols_pred != 1").
    reset_index().
    set_index("security")
    [["ols_pred"]].
    T)
```

```python
df_port = (df_model_wider.set_index(
    ["date", "security"]).
    mean(axis = 1).
    to_frame(name = "port").
    reset_index())
```

```python
(df_port.pivot(
    index = "date", columns = "security", values = "port").
    cumsum().
    plot())
```

```python
(df_port.drop(
    columns = ["date"]).
    groupby("security").
    agg(lambda x: x.mean() / x.std() * np.sqrt(252)).
    T)
```

# Single Period & Multi Period Returns with Trend and Mean Returns

```python
df_tsy_prep = (df_tsy.assign(
    date = lambda x: pd.to_datetime(x.date).dt.date).
    merge(right = df_dates, how = "inner", on = ["date"]).
    assign(security = lambda x: x.security.str.split("1").str[0].str.strip())
    [["date", "security", "day", "event", "PX_bps"]])
```

```python
df_two = (df_tsy_prep.assign(
    group    = lambda x: np.where(x.day <= 0, "before", "after"),
    approach = "two_stage"))
```

```python
df_tsy_grouped = (pd.concat([
    df_tsy_prep.assign(group = "all", approach = "one"),
    df_two]))
```

```python
def _get_trend_and_rtn(df: pd.DataFrame, window: float = 1) -> pd.DataFrame: 

    df_out = (df.assign(
        trend      = lambda x: x.PX_bps.ewm(span = 1, adjust = False).mean(),
        mean_bps   = lambda x: x.PX_bps.mean(),
        period_bps = lambda x: x.PX_bps.cumsum()).
        query("date == date.max()"))

    return df_out

df_trend = (df_tsy_grouped.assign(
    group_var = lambda x: x.group + x.approach + x.security + x.event.astype(str)).
    groupby("group_var").
    apply(_get_trend_and_rtn).
    reset_index(drop = True))
```

```python
def _lag_trend_and_mean_rtn(df: pd.DataFrame) -> pd.DataFrame: 

    df_out = (df.sort_values(
        "date").
        assign(
            lag_trend = lambda x: x.trend.shift(),
            lag_mean  = lambda x: x.mean_bps.shift()).
        dropna())

    return df_out

df_lagged = (df_trend.assign(
    group_var = lambda x: x.security + " " + x.approach + " " + x.group).
    groupby("group_var").
    apply(_lag_trend_and_mean_rtn).
    reset_index(drop = True))
```

```python
df_zscore_wider = df_zscore.pivot(index = "date", columns = "security", values = "lag_zscore")
```

```python
df_combined = (df_lagged.drop(
    columns = ["day", "trend", "mean_bps", "PX_bps"]).
    merge(right = df_zscore_wider, how = "inner", on = ['date']).
    fillna(0))
```

```python
def _get_pred(df: pd.DataFrame) -> pd.DataFrame: 

    df_tmp    = (df.assign(bin_bps = lambda x: np.where(x.period_bps > 0, 1, 0)))
    exog_cols = ["lag_trend", "lag_mean", "APUSISGF", "APUSSPGF", "APUSTYGF", "APUSXRGF", "BENLPFED"]

    ols_pred = (sm.OLS(
        endog = df_tmp.period_bps,
        exog  = sm.add_constant(df_tmp[exog_cols])).
        fit().
        fittedvalues)

    logit_pred = (sm.Logit(
        endog = df_tmp.bin_bps,
        exog  = sm.add_constant(df_tmp[exog_cols])).
        fit().
        fittedvalues)

    df_out = (df.assign(
        ols_pred   = ols_pred,
        logit_pred = logit_pred))

    return df_out

df_signal = (df_combined.assign(
    group_var = lambda x: x.security + " " + x.group + " " + x.approach).
    groupby("group_var").
    apply(_get_pred).
    reset_index(drop = True))
```

```python
df_combined_signal = (df_signal[
    ["security", "event", "group", "approach", "ols_pred", "logit_pred"]].
    merge(right = df_tsy_grouped, how = "inner", on = ["event", "security", "group", "approach"]).
    melt(id_vars = ["security", "event", "group", "approach", "date", "day", "PX_bps"]).
    assign(signal_rtn = lambda x: np.sign(x.value) * x.PX_bps))
```

```python
df_plot = (df_combined_signal.assign(
    plot_var = lambda x: x.approach + " " + x.variable))
```

```python
groups    = df_plot.plot_var.drop_duplicates().sort_values().to_list()
fig, axes = plt.subplots(ncols = 2, nrows = 2, figsize = (20,10))

renamer = {
    "one"       : "One Period",
    "two_stage" : "Two Period",
    "logit_pred": "Logistic",
    "ols_pred"  : "OLS"}

for group, ax in zip(groups, axes.flatten()): 

    approach, model = group.split(" ")
    
    (df_plot.query(
        "plot_var == @group")
        [["security", "date", "signal_rtn"]].
        groupby(["security", "date"]).
        head(1).
        rename(columns = {"security": ""}).
        pivot(index = "date", columns = "", values = 'signal_rtn').
        cumsum().
        plot(
            ax     = ax,
            ylabel = "Cuml. Basis Points",
            title  = renamer[approach] + " " + renamer[model]))

fig.suptitle("Cumulative Basis Points")
plt.tight_layout()
```

```python
(df_combined_signal[
    ["security", "variable", "signal_rtn", "approach"]].
    groupby(["security", "variable", "approach"]).
    agg(lambda x: x.mean() / x.std() * np.sqrt(252)).
    reset_index().
    pivot(index = ["variable", "approach"], columns = "security", values = "signal_rtn"))
```

# Single Period and Multi Period Cross Sectional Regression

```python
def _get_pred(df: pd.DataFrame) -> pd.DataFrame: 

    df_tmp    = (df.assign(bin_bps = lambda x: np.where(x.period_bps > 0, 1, 0)))
    exog_cols = ["lag_trend", "lag_mean", "APUSISGF", "APUSSPGF", "APUSTYGF", "APUSXRGF", "BENLPFED"]

    ols_pred = (sm.OLS(
        endog = df_tmp.period_bps,
        exog  = sm.add_constant(df_tmp[exog_cols])).
        fit().
        fittedvalues)

    logit_pred = (sm.Logit(
        endog = df_tmp.bin_bps,
        exog  = sm.add_constant(df_tmp[exog_cols])).
        fit().
        fittedvalues)

    df_out = (df.assign(
        ols_pred   = ols_pred,
        logit_pred = logit_pred))

    return df_out

df_combined_signal = (df_combined.assign(
    group_var = lambda x: x.group + " " + x.approach).
    groupby("group_var").
    apply(_get_pred).
    reset_index(drop = True))
```
